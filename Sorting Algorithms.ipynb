# **1. Insertion Sort**

```python
def insertion_sort(ls):
    for i in range(1,len(ls)):
        k=ls[i]
        j=i-1
        while j>=0 and ls[j]>k:
            ls[j+1]=ls[j]
            j-=1
        ls[j+1]=k
    return ls

ls=[64, 34, 25, 12, 22, 11, 90]
res=insertion_sort(ls)
print(res)
```

# **2. Bubble Sort**

```python
def bubble_sort(ls):
    n=len(ls)
    for i in range(n):
        s=False
        for j in range(n-i-1):
            if ls[j]>ls[j+1]:
                ls[j],ls[j+1]=ls[j+1],ls[j]
                s=True
        if not s:
            break
    return ls

ls=[64, 34, 25, 12, 22, 11, 90]
res=bubble_sort(ls)
print(res)
```

# **3. Heap Sort**

```python
def heapify(ls,n,i):
    m=i
    l=2*i+1
    r=2*i+2
    if l<n and ls[l]>ls[m]:
        m=l
    if r<n and ls[r]>ls[m]:
        m=r
    if m!=i:
        ls[i],ls[m]=ls[m],ls[i]
        heapify(ls,n,m)

def heap_sort(ls):
    n=len(ls)
    for i in range(n//2-1,-1,-1):
        heapify(ls,n,i)
    for i in range(n-1,0,-1):
        ls[i],ls[0]=ls[0],ls[i]
        heapify(ls,i,0)
    return ls

ls=[64, 34, 25, 12, 22, 11, 90]
res=heap_sort(ls)
print(res)
```

# **4. Selection Sort**

```python
def selection_sort(ls):
    n=len(ls)
    for i in range(n):
        m=i
        for j in range(i+1,n):
            if ls[j]<ls[m]:
                m=j
        ls[i],ls[m]=ls[m],ls[i]
    return ls

ls=[64, 34, 25, 12, 22, 11, 90]
res=selection_sort(ls)
print(res)
```

# **5. Merge Sort**

```python
def merge_sort(ls):
    if len(ls)<=1:
        return ls
    m=len(ls)//2
    l=merge_sort(ls[:m])
    r=merge_sort(ls[m:])
    return merge(l,r)

def merge(l,r):
    res=[]
    i=j=0
    while i<len(l) and j<len(r):
        if l[i]<=r[j]:
            res.append(l[i])
            i+=1
        else:
            res.append(r[j])
            j+=1
    res.extend(l[i:])
    res.extend(r[j:])
    return res

ls=[64, 34, 25, 12, 22, 11, 90]
res=merge_sort(ls)
print(res)
```

# **6. Quick Sort**

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quick_sort(left) + middle + quick_sort(right)
arr = [64, 34, 25, 12, 22, 11, 90]
q = quick_sort(arr)
print(q)
```

# **Knuth-Morris-Pratt (KMP) Algorithm:**

```python
def kmp_search(t,p):
    if not p or len(p)>len(t):
        return []
    lps=[0]*len(p)
    k,i=0,1
    while i<len(p):
        if p[i]==p[k]:
            k+=1
            lps[i]=k
            i+=1
        elif k!=0:
            k=lps[k-1]
        else:
            lps[i]=0
            i+=1

    res=[]
    i=j=0
    while i<len(t):
        if p[j]==t[i]:
            i+=1
            j+=1
            if j==len(p):
                res.append(i-j)
                j=lps[j-1]
        else:
            if j!=0:
                j=lps[j-1]
            else:
                i+=1
    return res

txt="ABABDABACDABABCABAB"
pat="ABABCABAB"
print(kmp_search(txt,pat))
```

# **Boyer-Moore Algorithm**

```python
def boyer_moore_search(t,p):
    n,m=len(t),len(p)
    if m==0 or n<m:
        return []

    s={c:m for c in set(t)}
    for k in range(m-1):
        s[p[k]]=m-1-k

    res=[]
    i=0
    while i<=n-m:
        j=m-1
        while j>=0 and p[j]==t[i+j]:
            j-=1

        if j<0:
            res.append(i)
            i+=1
        else:
            i+=s.get(t[i+m-1],m)

    return res

txt="ABABDABACDABABCABAB"
pat="ABABCABAB"
print(boyer_moore_search(txt,pat))
```
